# 📌 Consulta rápida

Una columna es **linealmente dependiente** si puede obtenerse exactamente como combinación lineal de otras columnas:

$$
v=\alpha_1u_1+\alpha_2u_2+\cdots+\alpha_ku_k.
$$

El **rango** indica cuántas columnas linealmente independientes contiene una matriz. Si

$$
\operatorname{rang}(A)<\text{número de columnas de }A,
$$

entonces las columnas no son todas linealmente independientes.

En Python:

```python
np.linalg.matrix_rank(A)
```

# **Actividad 5 - Análisis de indicadores mediante el rango 💻🐍**

Una empresa registra información de cuatro sucursales mediante tres indicadores operativos. La matriz de datos es:

$$A=
\begin{pmatrix}
10 & 20 & 30\\
15 & 30 & 45\\
20 & 40 & 60\\
25 & 50 & 75
\end{pmatrix}$$

Cada fila representa una sucursal y cada columna representa un indicador:

* columna 1: ventas base registradas;
* columna 2: indicador comercial derivado;
* columna 3: indicador operacional derivado.

El equipo de análisis desea saber si los tres indicadores entregan información independiente o si algunos de ellos son redundantes.

a) Cree la matriz $A$ en Python.

````python
import numpy as np

# Cada fila es una sucursal y cada columna un indicador
A = np.array([[10, 20, 30],
              [15, 30, 45],
              [20, 40, 60],
              [25, 50, 75]])

print("Matriz A:")
print(A)
print("Dimensión de A:", A.shape)
````

b) Calcule el rango de la matriz $A$.

````python
# El rango indica cuántas columnas linealmente independientes tiene A
rango_A = np.linalg.matrix_rank(A)

print("Rango de A:", rango_A)
print("Número de columnas de A:", A.shape[1])
````

c) Compare el resultado obtenido en el inciso anterior con el número de columnas de la matriz $A$. ¿Qué relación entre sus columnas permite explicar el rango obtenido? A partir de ello, interprete qué indica el resultado respecto del aporte de información de las tres columnas.

````python
# Columnas de A como vectores (cada una es un indicador)
c1 = A[:, 0]
c2 = A[:, 1]
c3 = A[:, 2]

# Verificamos si las columnas 2 y 3 son múltiplos de la columna 1
print("c2 / c1 =", c2 / c1)
print("c3 / c1 =", c3 / c1)
print("¿c2 = 2·c1?", np.array_equal(c2, 2 * c1))
print("¿c3 = 3·c1?", np.array_equal(c3, 3 * c1))
````

**Respuesta:**

El rango de $A$ es $1$, mientras que la matriz tiene $3$ columnas. Como

$$\operatorname{rang}(A)=1<3,$$

las columnas de $A$ **no** son linealmente independientes.

La relación entre las columnas explica este resultado: la segunda y la tercera columna son múltiplos de la primera,

$$c_2 = 2c_1, \qquad c_3 = 3c_1.$$

Por lo tanto, todas las columnas están generadas por un único vector ($c_1$), y solo hay **una** columna linealmente independiente.

**Interpretación:** aunque la empresa registra tres indicadores, estos contienen la información de una sola variable. El indicador comercial derivado corresponde al doble de las ventas base, y el indicador operacional derivado al triple de las ventas base, en todas las sucursales. Así, ambos indicadores derivados son **redundantes**: no aportan información nueva respecto de las ventas base registradas, y basta con conservar la columna 1 (o cualquiera de las tres) para describir la misma información.

# **Actividad 6 - Desempeño operacional 💻🐍**

Una empresa de soporte técnico registra, para cuatro equipos de trabajo, la cantidad de solicitudes resueltas y la cantidad de solicitudes que posteriormente debieron ser reabiertas. Además, utiliza un puntaje de desempeño operacional que se obtiene a partir de ambas variables.

Los registros disponibles son:
$$ c_1= \begin{pmatrix} 8\\ 10\\ 7\\ 12 \end{pmatrix}, \qquad c_2= \begin{pmatrix} 2\\ 3\\ 1\\ 4 \end{pmatrix}, \qquad c_3= \begin{pmatrix} 20\\ 24\\ 19\\ 28 \end{pmatrix}. $$
donde:
* $c_1$: solicitudes resueltas;
* $c_2$: solicitudes reabiertas;
* $c_3$: puntaje de desempeño operacional.

Se sabe que el puntaje $c_3$ se obtiene mediante una combinación lineal de $c_1$ y $c_2$:

$$ c_3 = \alpha c_1 + \beta c_2. $$


a) Considere los dos primeros registros y escriba las ecuaciones que deben satisfacer los coeficientes de $\alpha$ y $\beta$. Luego, organice el sistema en la forma matricial $Ax = b$ e indique qué presenta el vector $x$ en ese problema.

**Respuesta:**

Con los dos primeros registros ($c_3 = \alpha c_1 + \beta c_2$):

$$
\begin{cases}
8\alpha + 2\beta = 20\\
10\alpha + 3\beta = 24
\end{cases}
$$

En forma matricial $Ax=b$:

$$
\underbrace{\begin{pmatrix} 8 & 2\\ 10 & 3 \end{pmatrix}}_{A}
\underbrace{\begin{pmatrix} \alpha\\ \beta \end{pmatrix}}_{x}
=
\underbrace{\begin{pmatrix} 20\\ 24 \end{pmatrix}}_{b}
$$

El vector $x=(\alpha,\beta)$ contiene los **coeficientes de la combinación lineal**, es decir, el peso con que las solicitudes resueltas y las solicitudes reabiertas intervienen en el puntaje de desempeño.

b) Represente en Python la matriz $A$ y el vector $b$ obtenidos en el inciso anterior. Utilice `np.linalg.solve()` para determinar los valores de $\alpha$ y $\beta$.

````python
import numpy as np

# Sistema con los dos primeros registros
A = np.array([[8, 2],
              [10, 3]])
b = np.array([20, 24])

x = np.linalg.solve(A, b)
alpha, beta = x

# Se redondea para evitar pequeños errores de punto flotante
print("alpha =", round(alpha, 4))
print("beta  =", round(beta, 4))
````

c)	Escriba la combinación lineal obtenida e interprete, en el contexto del problema, el significado de los coeficientes $\alpha$ y $\beta$.

**Respuesta:**

La combinación lineal obtenida es

$$c_3 = 3c_1 - 2c_2.$$

**Interpretación:** el puntaje de desempeño operacional suma **3 puntos por cada solicitud resuelta** ($\alpha = 3$) y resta **2 puntos por cada solicitud reabierta** ($\beta = -2$). El coeficiente negativo indica que las reaperturas penalizan el desempeño, ya que reflejan solicitudes que no quedaron bien resueltas la primera vez.

d)	Compruebe si la relación obtenida también permite reproducir el puntaje de desempeño de los dos equipos restantes. A partir de los resultados, concluya si la misma regla permite construir todos los valores de $c_3$

````python
c1 = np.array([8, 10, 7, 12])
c2 = np.array([2, 3, 1, 4])
c3 = np.array([20, 24, 19, 28])

# Puntaje calculado con la regla obtenida
c3_calculado = alpha * c1 + beta * c2

print("c3 real:      ", c3)
print("c3 calculado: ", np.round(c3_calculado, 4))
print("Equipo 3:", round(c3_calculado[2], 4), "vs", c3[2])
print("Equipo 4:", round(c3_calculado[3], 4), "vs", c3[3])
print("¿Se cumple en todos los equipos?", np.allclose(c3, c3_calculado))
````

**Respuesta:**

Para los dos equipos restantes:

* Equipo 3: $3(7) - 2(1) = 21 - 2 = 19$ ✔
* Equipo 4: $3(12) - 2(4) = 36 - 8 = 28$ ✔

La relación $c_3 = 3c_1 - 2c_2$ reproduce exactamente el puntaje de los cuatro equipos. Por lo tanto, la misma regla permite construir todos los valores de $c_3$. Esto confirma que $c_3$ es combinación lineal de $c_1$ y $c_2$: el puntaje de desempeño es una variable **redundante**, porque no aporta información lineal nueva respecto de las solicitudes resueltas y reabiertas.

````python
import numpy as np

A = np.array([[1, 2, 4],
              [2, 1, 5],
              [0, 3, 3],
              [3, 0, 6],
              [4, 2, 10],
              [1, 4, 6]])

rango = np.linalg.matrix_rank(A)
print("Rango de A:", rango)
print("Cantidad de columnas:", A.shape[1])
````

````python
# Sistema con los dos primeros registros
M = np.array([[1, 2],
              [2, 1]])
y = np.array([4, 5])

alpha, beta = np.linalg.solve(M, y)
print("alpha =", round(alpha, 4))
print("beta  =", round(beta, 4))
````

**Respuesta:**

# 📌 Consulta rápida

**Producto punto**

$$
u\cdot v=\sum_{i=1}^{n}u_iv_i
$$

**Norma**

$$
\|u\|=\sqrt{\sum_{i=1}^{n}u_i^2}
$$

**Similitud coseno**

$$
\operatorname{sim}(u,v)=
\frac{u\cdot v}{\|u\|\|v\|}
$$

**Recuerda:** la distancia euclidiana permite comparar **cercanía de valores**, mientras que la similitud coseno permite comparar **similitud de patrón**.
